```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv('placement (2).csv')
df.head()
```

```python
df.shape
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('CGPA')
plt.ylabel('package')
```

```python
x = df.iloc[:,0:1]
y= df.iloc[:,-1]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=38)
```

```python
lr = LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
y_test.head()
```

```python
x_test.head()
```

```python
# Predict for all rows in x_test
y_pred = lr.predict(x_test)
```

```python
# Pass the first row as a DataFrame (keeps 2D shape automatically)
lr.predict(x_test.iloc[[2]])
```

```python
from sklearn.metrics import r2_score

# 1. R2 Score (returns a value up to 1.0; closer to 1 means better fit)
r2 = r2_score(y_test, y_pred)
print("R2 Score:", r2)
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel('CGPA')
plt.ylabel('package')
plt.plot(x_train,lr.predict(x_train),color = 'red')
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
```

```python
x_train_scaled = sc.fit_transform(x_train)
x_test_scaled = sc.transform(x_test)
```

```python
lr.fit(x_train_scaled,y_train)
```

```python
y_pred_new = lr.predict(x_test_scaled)
```

```python
from sklearn.metrics import r2_score

# 1. R2 Score (returns a value up to 1.0; closer to 1 means better fit)
r2_new = r2_score(y_test, y_pred_new)
print("R2 Score:", r2_new)
```

```python
from sklearn.model_selection import cross_val_score

scores = cross_val_score(LinearRegression(), x, y, cv=5, scoring='r2')
print("5-Fold Cross-Validated R2:", scores.mean())
```
