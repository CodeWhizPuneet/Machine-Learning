# multivariate analysis

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
titanic = pd.read_csv('train.csv')
flights = sns.load_dataset('flights')
iris = sns.load_dataset('iris')
```

```python
flights.head()
```

```python
iris.head()
```

```python
# Scatterplot (Numerical-Numerical)
```

```python
tips.head()
```

```python
sns.scatterplot(data=tips, x='total_bill', y='tip',hue='sex',style='smoker',size='size')
```

```python
# Barplot(Numerical - cateogorical)
```

```python
titanic.head()
```

```python
sns.barplot(data=titanic,x='Pclass',y='Age',hue='Sex')
```

```python
#Boxplot (Numerical - Cateogorical)
```

```python
sns.boxplot(data=titanic,x='Sex',y='Age',hue='Survived')
```

```python
sns.displot(data=titanic,x='Age',kind='kde',hue='Survived')
# sns.displot(data=titanic[titanic['Survived']==1],x='Age',kde=True,stat='density')
```

```python
# heatmap (cateogorical - cateogorical)
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
survival_rate = titanic.groupby("Pclass")["Survived"].mean()*100
print(survival_rate)
```

```python
(titanic.groupby("Sex")["Survived"].mean()*100)
```

```python
survival_rate.plot(kind="bar", color=["green","orange","red"])
plt.ylabel("Survival Rate (%)")
plt.xlabel("Passenger Class")
plt.title("Titanic Survival Rate by Class")
plt.show()
```

```python
#Clustermap (cateogorical - cateogorical)
```

```python
sns.clustermap(pd.crosstab(titanic['Parch'],titanic['Survived']))
```

```python
#Pairplot
```

```python
sns.pairplot(data=iris,hue='species')
```

```python
#Lineplot(Numerical - Numerical)
flights.head()
```

```python
flights_data = flights.groupby('year')['passengers'].sum().reset_index()
```

```python
sns.lineplot(data=flights_data,
             x='year',
             y='passengers')
```

```python
sns.clustermap(flights.pivot_table(values='passengers',index='month',columns='year'))
```
