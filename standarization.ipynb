```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('Social_Network_Ads.csv')
```

```python
df.sample(5)
```

```python
df.info()
```

```python
df = df[['Age','EstimatedSalary','Purchased']]
```

```python
df.head()
```

```python
#train_test_split
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(df.drop(['Purchased'],axis = 1),df['Purchased'],test_size=0.3,random_state=0)
```

```python
x_train.shape , x_test.shape
```

```python
#StandardScalar
!pip install scikit-learn
!pip install StandardScaler
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit(x_train)

x_train_scaled = scaler.transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
scaler.mean_
```

```python
x_train.head()
```

```python
x_train_scaled
```

```python
x_train_scaled = pd.DataFrame(x_train_scaled,columns = x_train.columns)
x_test_scaled = pd.DataFrame(x_test_scaled,columns = x_test.columns)
```

```python
x_train_scaled , x_test_scaled
```

```python
np.round(x_train.describe())
```

```python
np.round(x_train_scaled.describe())
```

```python
#Effect of scalling
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12,5))

sns.scatterplot(
    x=x_train['Age'],
    y=x_train['EstimatedSalary'],
    color='y',
    ax=ax[0]
)
ax[0].set_title('Before Scaling')

sns.scatterplot(
    x=x_train_scaled['Age'],
    y=x_train_scaled['EstimatedSalary'],
    color='red',
    ax=ax[1]
)
ax[1].set_title('After Scaling')

plt.tight_layout()
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12,5))

sns.kdeplot(x_train['Age'], ax=ax1)
sns.kdeplot(x_train['EstimatedSalary'], ax=ax1)
ax1.set_title('Before Scaling')

sns.kdeplot(x_train_scaled['Age'], ax=ax2)
sns.kdeplot(x_train_scaled['EstimatedSalary'], ax=ax2)
ax2.set_title('After Standard Scaling')

plt.show()
```

```python
#why scalling is important?
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr_scaled = LogisticRegression()
```

```python
lr.fit(x_train,y_train)
lr_scaled.fit(x_train_scaled,y_train)
```

```python
y_pred = lr.predict(x_test)
y_pred_scaled= lr_scaled.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score
```

```python
y_pred_accuracy = accuracy_score(y_test,y_pred)*100
y_pred_scaled_accuracy = accuracy_score(y_test,y_pred_scaled)*100
```

```python
print('Actual',y_pred_accuracy)
print('Scaled',y_pred_scaled_accuracy)
```
