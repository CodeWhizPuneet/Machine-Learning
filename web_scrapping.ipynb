```python
soup = BeautifulSoup(webpage, 'lxml')
```

```python
for i in soup.find_all('h2'):
    print(i.text.strip())
```

```python
company = soup.find_all('div',class_ = "companyCardWrapper")
```

```python
name=[]
rating=[]
review_count=[]
sector=[]
headquater=[]
highly_rated_for=[]
critically_rated_for=[]
for i in company:
    #extracting company name
    name.append(i.find('h2',class_="companyCardWrapper__companyName").text.strip()) 
    
    #extracting rating
    rating_tag = i.find('div', class_='rating_text rating_text--md')
    if rating_tag:
        rating.append(rating_tag.text.strip())
    else:
        rating.append(None)
        
    #extractin the review count
    review_count_tag = i.find('span', class_='companyCardWrapper__companyRatingCount')
    if review_count_tag:
        review_count.append(review_count_tag.text.strip().replace('(', '').replace(')', ''))
    else:
        review_count.append(None)

    #extraction of sector
    sector_tag = i.find('span',class_="companyCardWrapper__interLinking")
    if sector_tag:
        sector.append(sector_tag.text.split('|')[0].strip())
    else :
        sector.append(None)

    #extraction of city
    city_tag = i.find('span',class_="companyCardWrapper__interLinking")
    if city_tag:
        headquater.append(city_tag.text.split('|')[1].split('+')[0].strip())
    else :
        headquater.append(None)

    #extraction of highly rated for
    highly_rated_tag = i.find('span',class_="companyCardWrapper__ratingHeader--high")
    if highly_rated_tag:
        highly_rated_for.append(
            highly_rated_tag.find_parent('div')
            .find_next('span',class_="companyCardWrapper__ratingValues")
            .text.strip()
        )
    else :
        highly_rated_for.append(None)

    #extraction of critically rated for
    critically_rated_tag = i.find('span',class_="companyCardWrapper__ratingHeader--critical")
    if critically_rated_tag:
        critically_rated_for.append(
            critically_rated_tag.find_parent('div')
            .find_next('span',class_="companyCardWrapper__ratingValues")
            .text.strip()
        )
    else :
        critically_rated_for.append(None)
```

```python
d = {
    'company_name': name,
    'company_rating': rating,
    'review_count': review_count,
    'industry_sector': sector,
    'headquarters': headquater,
    'highly_rated_for': highly_rated_for,
    'critically_rated_for': critically_rated_for
}
```

```python
df = pd.DataFrame(d)
```

```python
final=pd.DataFrame()
for j in range(1,501):
    url='https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav&page={}'.format(j)
    headers = {'User-Agent':'Mozilla/5.0 (Windows NT 6.3; Win 64 ; x64) Apple WeKit /537.36(KHTML , like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
    webpage = requests.get(url, headers=headers).text
    soup = BeautifulSoup(webpage, 'lxml')
    company = soup.find_all('div',class_ = "companyCardWrapper")
    name=[]
    rating=[]
    review_count=[]
    sector=[]
    headquater=[]
    highly_rated_for=[]
    critically_rated_for=[]
    for i in company:
        #extracting company name
        name.append(i.find('h2',class_="companyCardWrapper__companyName").text.strip()) 
        
        #extracting rating
        rating_tag = i.find('div', class_='rating_text rating_text--md')
        if rating_tag:
            rating.append(rating_tag.text.strip())
        else:
            rating.append(None)
            
        #extractin the review count
        review_count_tag = i.find('span', class_='companyCardWrapper__companyRatingCount')
        if review_count_tag:
            review_count.append(review_count_tag.text.strip().replace('(', '').replace(')', ''))
        else:
            review_count.append(None)
    
        #extraction of sector
        sector_tag = i.find('span',class_="companyCardWrapper__interLinking")
        if sector_tag:
            sector.append(sector_tag.text.split('|')[0].strip())
        else :
            sector.append(None)
    
        # extraction of headquarters
        city_tag = i.find('span', class_="companyCardWrapper__interLinking")
        if city_tag:
            parts = city_tag.text.split('|')
        
            if len(parts) > 1:
                headquater.append(parts[1].split('+')[0].strip())
            else:
                headquater.append(None)
        else:
            headquater.append(None)
    
        #extraction of highly rated for
        highly_rated_tag = i.find('span',class_="companyCardWrapper__ratingHeader--high")
        if highly_rated_tag:
            highly_rated_for.append(
                highly_rated_tag.find_parent('div')
                .find_next('span',class_="companyCardWrapper__ratingValues")
                .text.strip()
            )
        else :
            highly_rated_for.append(None)
    
        #extraction of critically rated for
        critically_rated_tag = i.find('span',class_="companyCardWrapper__ratingHeader--critical")
        if critically_rated_tag:
            critically_rated_for.append(
                critically_rated_tag.find_parent('div')
                .find_next('span',class_="companyCardWrapper__ratingValues")
                .text.strip()
            )
        else :
            critically_rated_for.append(None)
            
    d = {
    'company_name': name,
    'company_rating': rating,
    'review_count': review_count,
    'industry_sector': sector,
    'headquarters': headquater,
    'highly_rated_for': highly_rated_for,
    'critically_rated_for': critically_rated_for
     }
    
    df = pd.DataFrame(d)
    final = pd.concat([final, df], ignore_index=True)
        
```
